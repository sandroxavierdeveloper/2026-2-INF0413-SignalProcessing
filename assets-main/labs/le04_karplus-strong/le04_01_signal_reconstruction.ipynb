```python
def create_sine_signal(M, A=1.0, phi= 0.0, fs=48e3, dur=1.0):
    '''
    M  : period (samples)
    A  : amplitude
    phi: phase
    dur: duration (seconds)
    fs : sampling frequency (Hz)
    '''

    M = int(M)
    w0 = 2 * np.pi / M
    f = fs / M # frequencia do sinal (Hz)
    N = int(dur * fs)
    n = np.arange(N)
    x = A * np.sin(w0 * n + phi)
    
    print(f'Frequência gerada: {f} Hz')
    return x, n
```

```python
M = 109
fs = 48000
dur = 2.0
A = 1.0
phi = 0.0

x, n = create_sine_signal(M, A, phi, fs, dur)
```

## Desafio

A partir da primitiva `SIGNALgenerate`, crie a função `SIGNALtone`, que permita criar um ***tom*** com ***3 segundos de duração*** e uma ***frequencia audível*** de $$\Large f=330\,(Hz)!$$

```python
def SIGNALtone(f=440, dur=1.0):
    fs = 48000
    N = int(dur * fs)
    n = np.arange(N)
    # f0 = (w0 / (2*pi)) * fs => w0 = 2*pi*f/fs
    w0 = 2 * np.pi * f / fs
    x = np.sin(w0 * n)
    return x, fs
```

```python
f = 330 # signal freq (Hz)
dur = 3.0 # duration (s)
```

```python
x, fs = SIGNALtone(f, dur)
```

```python
play(x, fs)
```

Como poderiamos comprovar que "deu certo"?

```python
n = np.arange(x.shape[0])
N = n.shape[0]

# Calculando o espectro usando FFT (Fast Fourier Transform)
import scipy.fftpack as fft
X = fft.fft(x)
fHz = fft.fftfreq(N, d=1/fs)
```

```python
Xmag = np.abs(X)
```

```python
fig, ax = plt.subplots()
ax.plot(fHz, Xmag, 'm')
ax.set_xlim([ 0, .02 * (fs / 2)])
figureFormat(ax, fig)
plt.show()
```

## Desafio - `osc2`

A partir da primitiva `SIGNALgenerate`, crie a função `SIGNALwaveform`, que admita tons com ***formas de onda*** (*waveforms*) distintas


```python
def SIGNALwaveform(f=440, dur=1.0):
    fs = 48000
    N = int(dur * fs)
    # M = fs / f
    M = int(np.round(fs / f))
    
    # Forma de onda aleatória (ruído branco num buffer M)
    buffer = np.random.randn(M)
    
    # Repete o buffer
    repeats = int(np.ceil(N / M))
    x = np.tile(buffer, repeats)[:N]
    
    return x, fs
```

```python
f = 330
dur = 2

y, fs = SIGNALwaveform(f, dur)
```

```python
play(y, fs)
```

```python
import scipy.fftpack as fft
N_y = len(y)
Y = fft.fft(y)
fHz_y = fft.fftfreq(N_y, d=1/fs)

Ymag = np.abs(Y)

fig, ax = plt.subplots()
ax.plot(fHz_y, Ymag, 'm')
ax.set_xlim([0, 0.02 * (fs / 2)])
plt.title('Espectro Harmônico da Forma de Onda Aleatória (Série de Fourier)')
plt.xlabel('Frequência (Hz)')
plt.grid(True)
plt.show()
```

```python
ny = np.arange(len(y))
fig, ax = plt.subplots()
ax.plot(ny, y, 'm')
ax.set_xlim([ 0, 10 * (fs / f) ]) # plota 10 períodos
plt.title('Visualização no tempo')
plt.xlabel('Amostras (n)')
plt.grid(True)
plt.show()
```
