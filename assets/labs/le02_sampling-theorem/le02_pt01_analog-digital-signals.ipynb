```python
# Resposta: Simulação de Carga e Descarga de um circuito RC
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros do circuito
V0 = 5.0      # Tensão da fonte (Volts)
R = 10e3      # Resistência (10 kOhms)
C = 100e-6    # Capacitância (100 uF)
tau = R * C   # Constante de tempo (RC) = 1.0 segundo

# Vetor de tempo de 0 a 5 segundos (5 vezes a constante de tempo)
t = np.linspace(0, 5, 500)

# 1. Equação da Carga do Capacitor: V(t) = V0 * (1 - e^(-t/tau))
v_carga = V0 * (1 - np.exp(-t / tau))

# 2. Equação da Descarga do Capacitor: V(t) = V0 * e^(-t/tau)
v_descarga = V0 * np.exp(-t / tau)

# Criando a figura com 2 gráficos lado a lado
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Gráfico de Carga
ax[0].plot(t, v_carga, 'b-', linewidth=2)
ax[0].set_title(f'Carga do Capacitor ($\\tau$ = {tau}s)')
ax[0].set_xlabel('Tempo (s)')
ax[0].set_ylabel('Tensão (V)')
ax[0].grid(True)

# Gráfico de Descarga
ax[1].plot(t, v_descarga, 'r-', linewidth=2)
ax[1].set_title(f'Descarga do Capacitor ($\\tau$ = {tau}s)')
ax[1].set_xlabel('Tempo (s)')
ax[1].set_ylabel('Tensão (V)')
ax[1].grid(True)

plt.tight_layout()
plt.show()
```

## Sinal de balística

<img src="./figures/balistica.png" alt="" width="500"/>

```python
# Resposta: Sinal de Balística (Lançamento de Projétil)
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros e Condições Iniciais do Lançamento
g = 9.81           # Aceleração da gravidade (m/s^2)
v0 = 50.0          # Velocidade inicial (m/s)
theta = 45.0       # Ângulo de lançamento em graus

# Convertendo o ângulo para radianos
theta_rad = np.radians(theta)

# Componentes Horizontal (x) e Vertical (y) da velocidade inicial
v0x = v0 * np.cos(theta_rad)
v0y = v0 * np.sin(theta_rad)

# Tempo total de voo (o projétil volta para y = 0)
t_voo = (2 * v0y) / g

# Vetor de tempo analógico de 0 até o impacto no chão
t = np.linspace(0, t_voo, 500)

# Equações horárias do movimento (Física)
x = v0x * t                        # Movimento Retilíneo Uniforme no eixo X
y = v0y * t - 0.5 * g * t**2       # Movimento Uniformemente Variado no eixo Y

# Plotando os gráficos
fig, ax = plt.subplots(1, 2, figsize=(12, 4))

# Gráfico 1: Altura pelo Tempo (O sinal contínuo em 1D)
ax[0].plot(t, y, 'g-', linewidth=2)
ax[0].set_title('Sinal Analógico: Altura vs Tempo')
ax[0].set_xlabel('Tempo (s)')
ax[0].set_ylabel('Altura (m)')
ax[0].grid(True)

# Gráfico 2: Trajetória 2D no espaço (Altura pela Distância)
ax[1].plot(x, y, 'm-', linewidth=2)
ax[1].set_title('Trajetória 2D do Projétil')
ax[1].set_xlabel('Distância (m)')
ax[1].set_ylabel('Altura (m)')
ax[1].grid(True)

plt.tight_layout()
plt.show()
```

## Conclusão: O Contínuo pelo Discreto
A simulação computacional de sistemas físicos (como circuitos RC e lançamentos parabólicos) é, em sua essência, a aplicação de ferramentas matemáticas de amostragem. Equações diferenciais contínuas (que governam a tensão no capacitor e a trajetória do projétil) são transformadas em equações de diferenças discretas, processadas no domínio digital amostra por amostra. Essa ponte entre os mundos analógico e digital é a base de todo o processamento de sinais modernos.
